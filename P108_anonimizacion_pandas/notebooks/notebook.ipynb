# P108: seudonimizacion y generalizacion con Pandas

El programa elimina identificadores directos, conserva los ultimos cuatro digitos de la tarjeta, genera identificadores aleatorios y agrupa edad, ciudad y ocupacion. El gasto y el orden de las filas se conservan porque asi lo requiere la actividad.


```python
from pathlib import Path
import sys

import pandas as pd

project = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / "P108_anonimizacion_pandas/src/main.py").is_file()
)
sys.path.insert(0, str(project))
from P108_anonimizacion_pandas.src.main import ACTIVITY_DIR, generalize_profiles

raw = pd.read_csv(ACTIVITY_DIR / "data/raw.csv")
auxiliary = pd.read_csv(ACTIVITY_DIR / "data/auxiliary.csv")
```

```python
def count_unique_matches(reference, candidates):
    sizes = reference.value_counts()
    return sum(sizes.get(tuple(profile), 0) == 1 for profile in candidates.to_numpy())

exact = ["age", "city", "occupation"]
pd.Series({
    "Perfiles auxiliares": len(auxiliary),
    "Coincidencias unicas antes": count_unique_matches(raw[exact], auxiliary[exact]),
    "Coincidencias unicas despues": count_unique_matches(
        generalize_profiles(raw), generalize_profiles(auxiliary)
    ),
})
```

El recuento solo mide coincidencias unicas por edad, lugar y ocupacion. Los cuatro digitos de la tarjeta, el gasto exacto y el orden pueden permitir otros cruces, por lo que esta entrega no garantiza anonimato.

Para regenerar el CSV, ejecuta desde la raiz del repositorio: `python -m P108_anonimizacion_pandas.src.main`. Cada ejecucion genera IDs nuevos.
